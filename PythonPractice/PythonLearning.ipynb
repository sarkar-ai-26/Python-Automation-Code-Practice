```python
Nested_dict = {"Data_1": {"Name" : "Yash", "Job" : "IT"},
                "Data_2" : {"Name" : "Raj", "Job" : "IT"},
                "Data_3" : {"Name" : "Vikas", "Job" : "IT"} }
for key, value in Nested_dict.items():
	print(f"Outer Key : {key}, Outer Value: {value}")
	for key_, value_ in value.items():
		print(f"Outer Key : {key_}, Outer Value: {value_}")
```

```python
#Map function map()
lists = [1,2,3,4,5]
print(list(map(lambda x:x**2,lists)))
# print(nlist)
```

```python
li1 = [1,2,3]
li2 = [4,6,8]

print(list(map(lambda x,y:x+y,li1,li2)))
```

```python
# map() to convert all strings values items in an iterable into a integer

print(list(map(int,['1','2','5','7'])))
```

```python
#convert all names in uppercase

names = ["yash","rohan","vikas"]
print(list(map(str.upper,names)))
```

```python
#Retrieve values inside a dictionary using map()

employess = [{"name":"yash","age":26},
             {"name":"raj","age":29}
             ]

def agefinder(listitem):
    return listitem["age"]

print(list(map(agefinder,employess)))
```
